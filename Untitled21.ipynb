````python
import time
import torch
import pandas as pd
from transformers import AutoTokenizer, AutoModelForCausalLM, TextStreamer


model_id = "unsloth/Llama-3.2-1B-Instruct"

print(f"Loading {model_id}...")
tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    torch_dtype=torch.bfloat16,
    device_map="auto"
)


data = [
    {"index": 10, "value": "Project Alpha"},
    {"index": 25, "value": "Project Beta"},
    {"index": 42, "value": "Project Gamma (Target)"},
    {"index": 55, "value": "Project Delta"},
    {"index": 99, "value": "Project Epsilon"}
]
df = pd.DataFrame(data)
dataset_str = df.to_string(index=False)
target_index = 42

prompts = {
    "Short": f"Dataset:\n{dataset_str}\n\nWhat is the value at index {target_index}?",

    "Descriptive": (
        f"Below is a tabular dataset containing index numbers and their corresponding values. "
        f"Please scan the data and find the exact value associated with index {target_index}.\n"
        f"Dataset:\n{dataset_str}\n\nValue:"
    ),

    "Long": (
        f"You are an advanced data extraction assistant. Your task is to act as a search indexer. "
        f"I will provide you with a string representation of a dataset where each row contains an "
        f"'index' and a 'value'. You must carefully read through all the rows, locate the row where "
        f"the index is exactly equal to {target_index}, and extract the string in the 'value' column. "
        f"Do not provide any extra explanation, just return the value.\n\n"
        f"Here is the dataset you need to process:\n"
        f"```text\n{dataset_str}\n```\n\n"
        f"Based on the instructions above, what is the value for the requested index?"
    )
}


class TimingStreamer(TextStreamer):
    """Custom streamer to capture Time-To-First-Token (TTFT) and End Time."""
    def __init__(self, tokenizer, **kwargs):
        super().__init__(tokenizer, **kwargs)
        self.start_time = None
        self.first_token_time = None
        self.end_time = None

    def put(self, value):

        if self.first_token_time is None:
            self.first_token_time = time.time()
        super().put(value)

    def end(self):
        self.end_time = time.time()
        super().end()

results = []

print("\n--- Starting Benchmark ---\n")

for prompt_type, prompt_text in prompts.items():
    print(f"\nEvaluating: {prompt_type} Prompt")
    print("-" * 40)


    messages = [{"role": "user", "content": prompt_text}]


    inputs = tokenizer.apply_chat_template(
        messages,
        add_generation_prompt=True,
        return_dict=True,
        return_tensors="pt"
    ).to(model.device)


    prompt_tokens_count = inputs["input_ids"].shape[1]


    streamer = TimingStreamer(tokenizer, skip_prompt=True)
    streamer.start_time = time.time()


    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=50,
            streamer=streamer,
            pad_token_id=tokenizer.eos_token_id,
            do_sample=False
        )


    generated_tokens_count = outputs.shape[1] - inputs["input_ids"].shape[1]


    prefill_time = streamer.first_token_time - streamer.start_time

    decode_time = streamer.end_time - streamer.first_token_time
    total_time = streamer.end_time - streamer.start_time

    results.append({
        "Scenario": prompt_type,
        "Prompt Tokens": prompt_tokens_count,
        "Gen Tokens": generated_tokens_count,
        "Processing Time (s)": round(prefill_time, 4),
        "Generation Time (s)": round(decode_time, 4),
        "Total Time (s)": round(total_time, 4)
    })


print("\n\n--- Benchmark Results ---")
results_df = pd.DataFrame(results)
print(results_df.to_string(index=False))
````
