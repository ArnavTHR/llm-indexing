```python
import os
import time
from google import genai
from google.genai import types
from google.colab import userdata


api_key = userdata.get('GEMINI_API_KEY')
client = genai.Client(api_key=api_key)
MODEL = "gemini-3.5-flash"
```

```python
import pandas as pd
df = pd.DataFrame({
    "Row_Index": range(200),
    "Server_ID": [f"SRV-{1000+i}" for i in range(200)],
    "Status": ["Online" if i % 3 else "Offline" for i in range(200)]
})

dataset = df.to_csv(index=False)

print(df)
```

```python
target = 173

prompts = {

    "Short":
    f"Find row {target}. Return Server_ID and Status.",


    "Medium":
    f"""
Look through the dataset and find Row_Index {target}.
Return the Server_ID and Status for that row only.
""",


    "Long":
    f"""
You are given a dataset containing server records.

Each record has:
- Row_Index
- Server_ID
- Status

Carefully search through the provided dataset and locate
the record where Row_Index equals {target}.

Once you find the correct record, return its Server_ID
and Status. Make sure you select exactly Row_Index {target}
and do not accidentally use a neighbouring row.

Return only the requested information.
""",


    "Descriptive":
    f"""
Act as a precise data retrieval system.

Your task is to inspect the dataset provided below and
retrieve information associated with Row_Index {target}.

First identify the row whose Row_Index exactly equals {target}.
Then read the Server_ID and Status values belonging to that
same row.

Accuracy is important. Do not infer values and do not select
rows {target-1} or {target+1}.

Return the Server_ID and Status.
"""
}
```

```python
results = []

for name, instruction in prompts.items():

    full_prompt = f"""
{instruction}

DATASET:
{dataset}
"""

    print("\n--------------------")
    print(name)
    print("--------------------")

    start = time.perf_counter()

    stream = client.models.generate_content_stream(
        model=MODEL,
        contents=full_prompt
    )

    first_token_time = None
    response_text = ""

    for chunk in stream:

        if chunk.text:

            if first_token_time is None:
                first_token_time = time.perf_counter()

            response_text += chunk.text

    end = time.perf_counter()

    ttft = first_token_time - start
    generation = end - first_token_time
    total = end - start

    results.append({
        "Prompt": name,
        "TTFT": ttft,
        "Generation": generation,
        "Total": total,
        "Response": response_text
    })

    print(response_text)
    print("TTFT:", round(ttft, 3), "s")
    print("Generation:", round(generation, 3), "s")
    print("Total:", round(total, 3), "s")
```
